**Sales Data Analysis**

```python
#Importing all necessary libraries
import pandas as pd
import  numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
#Load the CSV file
df=pd.read_csv('/content/Sales Data.csv')
```

```python
df.info()
```

```python
#Deleting columns Status and unnames1
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
#To check for null values
pd.isnull(df)
```

```python
#To check for null values
pd.isnull(df).sum()
```

```python
#To drop null values
df.dropna(inplace=True)
```

```python
pd.isnull(df).sum()
```

```python
df.shape
```

```python
#Changing Data Type for 'Amount' from FLOAT to INT
df['Amount']=df['Amount'].astype('int' )
```

```python
#Gives list of all columns
df.columns
```

```python
df[['Age','Orders','Amount']].describe()
```

**Gender**

```python
ax = sns.countplot(x = 'Gender', data = df,hue='Gender')
sns.set(rc={'figure.figsize':(7,5)})
for bars in ax.containers:
    ax.bar_label(bars)

ax.set_ylabel('Count of Orders Placed')



```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sg = df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sg,hue='Gender',palette='husl',legend=False)
```

From the above 2 graphs it is observed that women shop more than men and they spend more money for shopping as well.

**Age**

```python
ax=sns.countplot(x='Age Group',hue='Gender',data=df)
for bars in ax.containers:
  ax.bar_label(bars)

  ax.set_ylabel('Count of orders placed')
```

This graph shows that the maximum amount of orders are placed by the age group 26-35 in which women place more number of orders than men.

```python
s_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False )
sns.set(rc={'figure.figsize':(17,5) })
ax=sns.barplot(x='Age Group',y='Amount',data=s_age,hue='Age Group',palette='husl', legend=False)

for bars in ax.containers:
  ax.bar_label(bars,fmt='%.0f')
```

From the above graph, age 26-35 has spent the most amount of money on shopping.

**State**

```python
#Total number of orders from top 10 states
s_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(17,5) })
ax=sns.barplot(x='State',y='Orders',data=s_state,hue='State',palette='husl',legend=False)

for bars in ax.containers:
  ax.bar_label(bars)
```

```python
#Total Amount from top 10 states

# Aggregate and sort data
s_age = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

# Create bar plot using Seaborn only
sns.set(rc={'figure.figsize': (17,5)})  # Set figure size
ax = sns.barplot(x='State', y='Amount', data=s_age, hue='State', palette="husl", legend=False)

for bars in ax.containers:
    ax.bar_label(bars,fmt='%.0f')

```

The first graph shows the top 10 states with the highest number of orders placed, sorted in descending order. The second graph displays the top 10 states based on the total amount spent on orders. Notably, in the first graph, the third state from the right is Kerala, while in the second graph, it's Haryana. This indicates that although more orders were placed from Kerala than Haryana, the total amount spent on orders is higher in Haryana than in Kerala.

**Marital Status**

```python
ax=sns.countplot(x='Marital_Status',data=df,hue='Marital_Status',legend=False)
sns.set(rc={'figure.figsize':(7,5)})


for bars in ax.containers :
  ax.bar_label(bars,fmt='%.0f')

  ax.set_xticks([0, 1])  # Explicitly set tick positions
ax.set_xticklabels(['Married', 'Unmarried'])
```

```python
s_maritalstatus=df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(7,5)})
ax=sns.barplot(x='Marital_Status',hue='Gender',y='Amount',data=s_maritalstatus)


for bars in ax.containers :
  ax.bar_label(bars,fmt='%.0f')

  ax.set_xticks([0, 1])  # Explicitly set tick positions
ax.set_xticklabels(['Married', 'Unmarried'])


```

From the above graphs we can see that most of the buyers are married and of those, most are women and they have high purchasing power.

**Occupation**

```python
ax=sns.countplot(x='Occupation',data=df,hue='Occupation',legend=False)
sns.set(rc={'figure.figsize':(20,5)})

for bars in ax.containers:
  ax.bar_label(bars)
  ax.set_ylabel("Count of Orders")
```

```python

s_occupation=df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.barplot(x='Occupation',y='Amount',data=s_occupation,hue='Occupation')
for bars in ax.containers:
  ax.bar_label(bars,fmt='%.0f')
  ax.set_ylabel('Amount')
```

From the above graphs we can see that most of the buyers are from IT, Healthcare, and Aviation sector.

**Product Category**

```python
ax=sns.countplot(x='Product_Category',data=df)
sns.set(rc={'figure.figsize':(30,10)})

for bars in ax.containers:
  ax.bar_label(bars)
  ax.set_ylabel("Count of Orders")
```

```python
s_Product_Category=df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(30,10)})
ax=sns.barplot(x='Product_Category',y='Amount',data=s_Product_Category,hue='Product_Category',legend=False)
for bars in ax.containers:
  ax.bar_label(bars,fmt='%.0f')
  ax.set_ylabel("Amount")
```

```python
s_Product_Category=df.groupby(['Product_ID'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(30,10)})
ax=sns.barplot(x='Product_ID',y='Amount',data=s_Product_Category,hue='Product_ID',legend=False)
for bars in ax.containers:
  ax.bar_label(bars,fmt='%.0f')
  ax.set_ylabel("Amount")
```
